## **1. Creating and Manipulating Strings**

Strings are fundamental data types in Python that store sequences of characters. Understanding how to
create and manipulate strings is essential for handling textual geospatial data effectively

### 1.1 Creating Strings 

Strings can be created by enclosing characters in single **quotes ( ' ), double quotes ( " ), or triple quotes
( ''' or """ )** for multi-line strings.

```python
# Different ways to create strings
location_name = "Mount Everest"  # Using double quotes
country = 'Nepal'  # Using single quotes
description = """Mount Everest is the highest peak in the world, located in the Himalayas."""  # Multi-line string

print(f"Location : {location_name}")
print(f"Country : {country}")
print(f"Description : {description}")
```

### 1.2 String Concatenation

Concatenation means joining strings together. This is useful for building location descriptions, file paths, or combining separate pieces of geographic information. Concatenation is done using the **(+)** operator.

```python
# Basic concatenation using the + operator
location_full = location_name + ", " + country
print(f"Full Location : {location_full}")
```

```python
# Building a file path
data_folder = "geographic_data"
filename = "mountain_peaks.csv"
filepath = data_folder + "/" + filename
print(f"File path : {filepath}")
```

### 1.3 String Repetition

The **(*)** operator allows you to repeat strings, which can be useful for creating separators, borders, or
formatting output.

```python
# Create visual separators
separator = "-" * 30
print(separator)
print("Geographic Data Report")
print(separator)
```

```python
# Create visual separators
tab_space = " " * 4
print(f"Location:{tab_space}{location_name}")
print(f"Elevation:{tab_space}8,848 meters")
```

### 1.4 String Length and Basic Properties


You can get information about strings using built-in functions. For example, you can get the length of a string using the **len()** function. You can also check if a string contains only letters using the **isalpha()** method. To check if a string contains only digits, you can use the **isdigit()** method. Below is an example of how to use these functions.

```python
# Get the length of a string
location_length = len(location_name)
print(f"The location name '{location_name}' has {location_length} characters")
```

```python
# Check if a string contains only letters
city_name = "SanFrancisco"
print(f"Is '{city_name}' alphabetic? {city_name.isalpha()}")
```

```python
# Check if a string contains only digits (useful for coordinate validation)
zip_code = "94102"
print(f"Is '{zip_code}' numberic? {zip_code.isdigit()}")
```

Strings are often built dynamically based on variable data, which is common when working with

## **2. String Methods for Geospatial Data**

Python provides powerful built-in methods that allow you to transform and manipulate strings. These
methods are particularly useful in geospatial programming for cleaning data, standardizing formats, and
extracting information from text-based geographic data.

### 2.1 Case Convertion Methods

Converting between uppercase and lowercase is essential for standardizing geographic data, especially
when working with place names from different sources. The **upper() , lower() , title() , and
capitalize()** methods are particularly useful for this purpose.

```python
# Case conversion examples
mountain_name = "Mount Everest"

# Convert to different cases
print(f"Original: {mountain_name}")
print(f"Uppercase: {mountain_name.upper()}")
print(f"Lowercase: {mountain_name.lower()}")
print(f"Title case: {mountain_name.title()}")
print(f"Capitalize: {mountain_name.capitalize()}")
```

### 2.2 Whitespace Removal Methods

Geographic data often comes with unwanted spaces that need to be cleaned. The strip methods help
remove these spaces. There are three methods: **strip() , lstrip() , and rstrip()** . The **strip()**
method removes both leading and trailing spaces, while the **lstrip()** method removes leading spaces
and the **rstrip()** method removes trailing spaces. Here is an example:

```python
# Whitespace removal examples
messy_location = " San Francisco "
messy_left = " Los Angeles"
messy_right = "Chicago "

print(f"Original: '{messy_location}'")
print(f"strip(): '{messy_location.strip()}'")
print(f"lstrip(): '{messy_left.lstrip()}'")
print(f"rstrip(): '{messy_right.rstrip()}'")
```

### 2.3 String Replacement

The **replace()** method allows you to substitute parts of strings, which is useful for correcting data or
updating information:

```python
# Basic replacement
location = "Mount Everest, Nepal"
update_location = location.replace("Everest", "Kilimanjaro")
print(f"Original: {location}")
print(f"Updated: {update_location}")
```

```python
# Replace multiple occurrences
path_string = "data/raw_data/geographic_data/raw_data/points.csv"
clean_path = path_string.replace("raw_data/", "")
print(f"Original path: {path_string}")
print(f"Clean path: {clean_path}")
```

### 2.4 String Splitting

The **split()** method breaks strings into lists, which is extremely useful for parsing structured
geographic data. For example, you might have a string that contains a list of city names separated by
commas, and you want to split it into a list of city names. The split() method takes a separator as an
argument, and returns a list of strings. The default separator is a space, but you can specify a different
separator if needed. The example below shows how to split a string into a list of city names using the
split() method with a comma as the separator:               

```python
# Basic splitting
location_full = "Mount Everest, Nepal, Asia"
location_parts = location_full.split(", ")
print(f"Orignial: {location_full}")
print(f"Split into parts: {location_parts}")
```

```python
# Extract individual components
mountain, country, continent = location_parts
print(f"Mountain: {mountain}")
print(f"Country: {country}")
print(f"Continent: {continent}")
```

The example below shows how to split a string into a list of coordinates:

```python
# Splitting coordinate strings
coordinate_string = "40.7128,-74.0060"
lat_str, lon_str = coordinate_string.split(",")
latitude = float(lat_str)
longitude = float(lon_str)
print(f"Parse Coordinates: Lat = {latitude}, Lon = {longitude}")
```

The example below shows how to split a file path into a list of path components using the split()
method with a forward slash as the separator:

```python
# Splitting file paths
file_path = "data/geographic/cities/world_cities.csv"
path_component = file_path.split("/")
print(f"Path component: {path_component}")
print(f"Filename: {path_component[-1]}")
print(f"Folderdata: {path_component[0]}")
```

### 2.5 String Joining

The **join()** method combines lists of strings into single strings, which is the reverse of splitting. For
example, you might have a list of city names, and you want to join them into a single string with a comma
and a space between each city name. To use the join() method, you need to provide a string as the
separator. The example below shows how to join a list of city names into a single string with a comma
and a space between each city name:

```python
# Basic joining
city_names = ["San Francisco", "Florida", "Philadelphia"]
city_names = ", ".join(city_names)
print(f"Joined city names: {city_names}")
```

The example below shows how to join a list of file paths into a single file path using the join() method
with a forward slash as the separator:

```python
# Creating file paths
file_path = ["data", "geographic", "cities", "world_citis.cvs"]
full_path = "/".join(file_path)
print(f"Full path: {full_path}")
```

The example below shows how to join a list of coordinates into a string:

```python
# Practical example: creating coordinate strings
coordinates = ["40.7128", "-74.0060"]
coordinates_str = ",".join(coordinates)
print(f"Coordinates String: {coordinates_str}")
```

## **3. String Formatting**

String formatting allows you to create strings that incorporate variable values, which is essential for
generating reports, creating file names, building queries, and displaying geographic information. Python
provides several powerful formatting methods that make it easy to create well-structured text .

### 3.1  F-String Formatting (Recommended)

F-strings (formatted string literals) are the most modern and readable way to format strings in Python.
They’re particularly useful for creating geographic reports and displaying coordinate information. To use
f-strings, you can use the f prefix before the string, and then use curly braces to insert the variable
values. The example below shows how to use f-strings to format a string with a variable value:

```python
# Basic f-string formatting with geographic data
location = "Mount Everest"
latitude = 27.9881
longitude = 86.9250
elevation = 8848

# Simple variable insertion
location_info = f"Location: {location}"
print(location_info)

# Multiple variables
Coordinates = f"Coordinates: ({latitude}, {longitude})"
print(Coordinates)

# Complete geographic summary
summary = f"{location} is located at {latitude}°N, {longitude}°E with an elevation of {elevation} meters"
print(summary)
```

### 3.2 Formatting Numbers in Strings

When working with geographic data, you often need to control how numbers are displayed. For example,
you might want to display a coordinate with a certain number of decimal places, or a number with a
certain number of significant figures. To do this, you can use f-strings to format numbers. The format
specifier is a colon : followed by the number of decimal places or significant figures. The example below
shows how to format a number with a certain number of decimal places:

```python
# Controlling decimal places for coordinates
precise_lat = 40.712776
precise_lon = -74.005974

# Round to different decimal places
coords_2_places = f"Coordinates: ({precise_lat:.2f}, {precise_lon:.2f})"
coords_4_places = f"Coordinates: ({precise_lat:.4f}, {precise_lon:.4f})"

print (coords_2_places)
print(coords_4_places)
```

```python
# Adding thousands separators for large numbers
population = 8336817
area_sqkm = 783.8

formatted_stats = f"NYC Population: {population:,} people, Area: {area_sqkm:.1f} km²"
print(formatted_stats)
```

### 3.3 Legacy Fomatting Methods

While f-strings are preferred, you should be familiar with older formatting methods that you might
encounter in older code. The example below shows how to format a number with a certain number of
decimal places using the **format()** method. Instead of using curly braces, you use curly braces with a number inside to indicate the position of the variable to be formatted. The example below shows how to
format a number with a certain number of decimal places:

```python
# Using .format() method
location = "San Francisco"
lat = 37.7749
lon = -122.4194

# Basic format method
formatted_1 = "Location: {} at coordinates ({}, {})".format(location, lat, lon)
print(formatted_1)

# With positional arguments
formatted_2 = "Location: {0} at coordinates ({1}, {2})".format(location, lat, lon)
print(formatted_2)

# With named arguments
formatted_3 = "Location: {name} at coordinates ({latitude}, {longitude})".format(name=location, latitude=lat, longitude=lon)
print(formatted_3)
```

### 3.5 Practical Formatting Examples

Here are some common geospatial formatting scenarios. For example, you might want to save a coordinate
as a string with a certain number of decimal places in a file with timestamp in the filename. The example
below shows how to do this:

```python
# Creating file names with timestamps and coordinates
import datetime

current_time = datetime.datetime.now()
survey_lat = 45.3311
survey_lon = -121.7113

filename = f"survey_{current_time.strftime('%Y%m%d')}_{survey_lat:.4f}N_{abs(survey_lon):.4f}W.csv"
print(f"Generated filename: {filename}")
```

```python
# Creating Well-Known Text (WKT) representations
wkt_point = f"POINT({survey_lon} {survey_lat})"
print(f"WKT Point: {wkt_point}")
```

The example below shows how to build a SQL query with formatting

```python
# Building SQL queries with formatting
table_name = "cities"
min_population = 1000000
region = "North America"

sql_query = f"""SELECT name, latitude, longitude
FROM {table_name} 
WHERE population > {min_population:,}
AND region = '{region}'"""

print(f"Generated SQL Query:")
print(sql_query)
```

Choosing the right string operation for your geospatial task is important for writing efficient and main￾tainable code:

### 4.1 When to Use Each Operation

**Use concatenation ( + or join() ) when**
1. Building file paths or URLs
2. Combining location names with additional information
3. Creating simple text outputs

**Use string methods ( strip() , replace() , split() ) when:**
1. Cleaning messy geographic data
2. Standardizing place names or coordinate formats
3. Breaking apart structured text data

**Use formatting (f-strings) when:**
1. Creating reports or displaying geographic information
2. Building dynamic queries or file names
3. Controlling number precision in coordinate displays

### Exercise 1: Manipulating Geographic Location Strings

Instruction:
1. Create a string that represents the name of a geographic feature (e.g., "Amazon River" ).
2. Convert the string to lowercase and then to uppercase.
3. Concatenate the string with the name of the country (e.g., "Brazil" ) to create a full location name.
4. Repeat the string three times, separating each repetition with a dash ( - ).

```python
# ANSWER
geographic_feature = "Amazon River"
country_name = "Brazil"

# Convert to lowercase and uppercase 
lowercase = geographic_feature.lower()
uppercase = geographic_feature.upper()

print(f"Lowercase: {lowercase}")
print(f"Uppercase: {uppercase}")

#  Concatenate the string with the name of the country
full_location = geographic_feature + ", " + country_name
print(f"Full Location Name: {full_location}")

# Repeat the string three times, separating each repetition with a dash ( - )
repeat_str = "-".join([full_location] * 3)
print(repeat_str)
```

### Exercise 2: Extracting and Formatting Coordinates

Instruction:
1. Given a string with the format "latitude, longitude" (e.g., "40.7128N, 74.0060W" ), extract the numeric values of latitude and longitude.
2. Convert these values to floats and remove the directional indicators ( N , S , E , W ).
3. Format the coordinates into a POINT WKT string (e.g., "POINT(-74.0060 40.7128)" ).


```python
# ANSWER
coordinates = "40.7128N, 74.0060W"

# Given a string with the format "latitude, longitude" (e.g., "40.7128N, 74.0060W" ), extract the numeric values of latitude and longitude
coordinates_parts = coordinates.split(", ")
latitude_str, longitude_str = coordinates_parts

print(f"Lat: {latitude_str}, Lon: {longitude_str}")

# Convert these values to floats and remove the directional indicators ( N , S , E , W )
latitude = float(latitude_str.replace("N", ""))
longitude = -float(longitude_str.replace("W",""))

print(f"Coordinates: ({latitude}, {longitude})")

# Format the coordinates into a POINT WKT string
wkt_point = f"POINT({longitude} {latitude})"
print(f"WKT Point: {wkt_point}")


```

### Exercise 4: String Normalization and Cleaning

Instruction:
1. Given a list of city names with inconsistent formatting (e.g., [" new york ", "Los ANGELES", " CHICAGO"] ), normalize the names by:
   - Stripping any leading or trailing whitespace.
   - Converting them to title case (e.g., "New York" , "Los Angeles" , "Chicago" ).
2. Ensure that the output is a clean list of city names.


```python
# ANSWER
city_names = [" new york ", "Los ANGELES", " CHICAGO"]

NY = city_names[0].strip().title()
LA = city_names[1].title()
CH = city_names[2].lstrip().title()

clean_city_names = NY + ', ' + LA + ', ' + CH

list_city = clean_city_names.split(", ")
print(f"List City Names: {list_city}")


```

```python
# ALT
city_names = [" new york ", "Los ANGELES", " CHICAGO"]

clean_city_names = [city.strip().title() for city in city_names]

print(f"List City Names: {clean_city_names}")
```

### Exercise 5: Parsing and Extracting Address Information

Instruction:
1. Given a string in the format "Street, City, Country" (e.g., "123 Main St, Springfield, USA" ), write a function that parses the string into a dictionary with keys street , city , and country .
2. The function should return a dictionary like {"street": "123 Main St", "city": "Springfield", "country": "USA"} .

```python
# ANSWER 
address = "123 Main St, Springfield, USA"
address_str = address.split(", ")
street, city, country = address_str
address_info = {
     "street": street,
     "city": city,
     "country": country
}
print(address_info)
```

```python
# ALT
def parse_address(address):
    street, city, country = address.split(", ")

    address_info = {
        "street": street,
        "city": city,
        "country": country
    }

    return address_info


address = "123 Main St, Springfield, USA"

result = parse_address(address)

print(result)
```
